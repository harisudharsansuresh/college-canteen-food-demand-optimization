```python
import numpy as np
import pandas as pd

print("Libraries imported successfully!")
```

```python
# Number of days to simulate
num_days = 1000

# Possible customer demand
demand_values = [150, 180, 200, 220, 250]

# Probability of each demand
probabilities = [0.10, 0.20, 0.40, 0.20, 0.10]

# Set random seed for reproducibility
np.random.seed(42)

# Generate random demand for 1000 days
demand = np.random.choice(
    demand_values,
    size=num_days,
    p=probabilities
)

# Create Day numbers
days = np.arange(1, num_days + 1)

# Create a DataFrame
df = pd.DataFrame({
    "Day": days,
    "Demand": demand
})

# Display first 10 rows
print(df.head(10))
```

```python
# Save dataset as CSV
df.to_csv("demand_data.csv", index=False)

print("Dataset saved successfully!")
```

```python
# Check the number of rows and columns
print("Dataset shape:", df.shape)

# Check average demand
print("Average demand:", df["Demand"].mean())

# Check demand distribution
print(df["Demand"].value_counts().sort_index())
```

```python
import matplotlib.pyplot as plt

# Plot demand distribution
df["Demand"].value_counts().sort_index().plot(
    kind="bar",
    title="College Canteen Daily Demand Distribution"
)

plt.xlabel("Number of Meals Demanded")
plt.ylabel("Number of Days")
plt.show()
```

```python
# Define canteen parameters

selling_price = 70
preparation_cost = 40
leftover_value = 5
shortage_penalty = 10
max_capacity = 300

# Function to calculate profit

def calculate_profit(Q, D):

    meals_sold = min(Q, D)

    leftover = max(Q - D, 0)

    shortage = max(D - Q, 0)

    revenue = meals_sold * selling_price

    salvage = leftover * leftover_value

    production_cost = Q * preparation_cost

    penalty = shortage * shortage_penalty

    profit = revenue + salvage - production_cost - penalty

    return profit
```
